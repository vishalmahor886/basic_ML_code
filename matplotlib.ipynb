```python
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline        #don't need to write plt.show
```

```python
x=np.arange(0,10)
y=np.arange(11,21)
```

A scatter plot in Matplotlib is used to show the relationship between two numerical variables. Each point represents one data value.


What is a Scatter Plot?

Displays data as points (dots)

Used to find patterns, trends, correlation

Common in data analysis & machine learning

```python
#ploting
#plt scatter

plt.scatter(x,y, c='b')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Scatter Plot')
plt.savefig('scatter_plot.png')
```

```python
from matplotlib.lines import lineStyles
plt.plot(x,y,'r')
plt.grid(True)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Line Plot')
```

```python
#subplot
plt.subplot(2,2,1)
plt.plot(x,y,'r*')
plt.subplot(2,2,2)
plt.plot(x,y,'g*')
plt.subplot(2,2,3)
plt.plot(x,y,'b*')
```

```python


x1 = [1, 2, 3, 4]
y1 = [10, 20, 30, 40]
y2 = [40, 30, 20, 10]

plt.subplot(1, 2, 1)
plt.plot(x1, y1)
plt.title("Plot 1")

plt.subplot(1, 2, 2)
plt.plot(x1, y2)
plt.title("Plot 2")

plt.show()
```

```python
fig, axs = plt.subplots(2, 2)

axs[0, 0].plot(x1, y1)
axs[0, 0].set_title("Plot 1")

axs[0, 1].plot(x1, y2)
axs[0, 1].set_title("Plot 2")

axs[1, 0].scatter(x1, y1)
axs[1, 0].set_title("Scatter")

axs[1, 1].bar(x1, y1)
axs[1, 1].set_title("Bar")

plt.show()
```

```python
x=np.arange(0,4*np.pi,0.1)
y=np.sin(x)
plt.plot(x,y)
plt.title("sin wave")
```

```python
fig, axe = plt.subplots(2,2)
axe[0,0].plot(x,np.sin(x),'r--')
axe[0,0].set_title("sin wave")

axe[0,1].plot(x,np.cos(x),'b')
axe[0,1].set_title("cos wave")

axe[1,0].plot(x,np.tan(x))
axe[1,0].set_title("tan wave")
```

```python

categories = ["A", "B", "C", "D"]
values = [10, 20, 15, 25]

plt.bar(categories, values)
plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Simple Bar Chart")
plt.show()
```

```python
plt.barh(categories, values)
plt.show()
```

```python
plt.bar(categories, values, color="green", width=0.5)
plt.show()
```

```python
bars = plt.bar(categories, values)

for bar in bars:
    plt.text(bar.get_x() + bar.get_width()/2,
             bar.get_height(),
             bar.get_height(),
             ha="center", va="bottom")

plt.show()
```

```python
import numpy as np

labels = ["A", "B", "C", "D"]
x = np.arange(len(labels))

y1 = [10, 20, 15, 25]
y2 = [12, 18, 20, 22]

plt.bar(x - 0.2, y1, width=0.4, label="Group 1")
plt.bar(x + 0.2, y2, width=0.4, label="Group 2")

plt.xticks(x, labels)
plt.legend()
plt.show()
```

Histogram

A histogram is a graph showing frequency distributions.

It is a graph showing the number of observations within each given interval.

Example: Say you ask for the height of 250 people, you might end up with a histogram like this

```python
data = [10, 12, 15, 18, 20, 22, 25, 28, 30]

plt.hist(data, bins=5)
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Histogram")
plt.show()
```

```python
x = np.random.normal(170, 10, 250)

plt.hist(x, density=True,  color="red", edgecolor="black")
plt.xlabel("Height")
plt.ylabel("Frequency")
plt.show()
```

PIE CHART

A pie chart in Matplotlib is used to show proportions or percentages of a whole.

What is a Pie Chart?

Circular chart divided into slices

Each slice represents a part of the total

Best when the number of categories is small

```python
y = np.array([35,25,25,15])
mylabels=["apple","bananan","watermelon","kiwi"]
plt.pie(y,labels=mylabels)
```

```python
plt.pie(y,labels=mylabels, startangle=90)
```

```python
myexplode = [0.2, 0.1, 0, 0]

plt.pie(y,labels=mylabels, startangle=90,explode=myexplode)
```

```python
mycolors = ["black", "hotpink", "b", "#4CAF50"]
plt.pie(y,labels=mylabels, startangle=90,explode=myexplode, autopct='%1.1f%%', colors=mycolors)
plt.legend(title="Four Fruits:")
```
