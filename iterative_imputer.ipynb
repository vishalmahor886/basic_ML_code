```python
import pandas as pd
import numpy as np
```

```python
url = 'https://raw.githubusercontent.com/campusx-official/100-days-of-machine-learning/main/day40-iterative-imputer/50_Startups.csv'
df = pd.read_csv(url, usecols=['R&D Spend','Administration','Marketing Spend','Profit'])
```

```python
df = df/10000
```

```python
df.sample(5)
```

```python
df.isnull().mean()*100
```

```python
df = df.iloc[:,0:-1]
df
```

```python
df.iloc[1,0] = np.nan
df.iloc[3,1] = np.nan
df.iloc[-1,-1] = np.nan
```

```python
df.isnull().mean()*100
```

```python
df.sample(5)
```

```python
# Step 1 - Impute all missing values with mean of respective col

df0 = pd.DataFrame()

df0['R&D Spend'] = df['R&D Spend'].fillna(df['R&D Spend'].mean())
df0['Administration'] = df['Administration'].fillna(df['Administration'].mean())
df0['Marketing Spend'] = df['Marketing Spend'].fillna(df['Marketing Spend'].mean())
```

```python
# 0th Iteration
df0
```

```python
# Remove the col1 imputed value
df1 = df0.copy()

df1.iloc[1,0] = np.nan

df1
```

```python
# Use first 3 rows to build a model and use the last for prediction

X = df1.iloc[[0,2,3,4],1:3]
X
```

```python
y = df1.iloc[[0,2,3,4],0]
y
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[1,1:].values.reshape(1,2))
```

```python
df1.iloc[1,0] = 15.68
```

```python
df1
```

```python
# Remove the col2 imputed value

df1.iloc[3,1] = np.nan

df1
```

```python
# Use last 3 rows to build a model and use the first for prediction
X = df1.iloc[[0,1,2,4],[0,2]]
X
```

```python
y = df1.iloc[[0,1,2,4],1]
y
```

```python
lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[3,[0,2]].values.reshape(1,2))
```

```python
df1.iloc[3,1] = 11.38
```

```python
df1
```

```python
# Remove the col3 imputed value
df1.iloc[4,-1] = np.nan

df1
```

```python
# Use last 3 rows to build a model and use the first for prediction
X = df1.iloc[0:4,0:2]
X
```

```python
y = df1.iloc[0:4,-1]
y
```

```python
lr = LinearRegression()
lr.fit(X,y)
lr.predict(df1.iloc[4,0:2].values.reshape(1,2))
```

```python
df1.iloc[4,-1] = 36.28
```
