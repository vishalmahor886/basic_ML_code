```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(np.arange(0,20).reshape(5,4),
                  index=["row1","row2", "row3", "row4", "row5"],
                  columns= ["col1","col2","col3","col4"])
```

```python
print(df)
```

```python
df.to_csv('Test.csv')   #take data into csv file
```

```python
#accesing the elements
#1. .loc    #2. .iloc

## using row
df.loc['row1']
type(df.loc['row1'])            # datafram is colection of the series
print(df.loc["row1","col2"])    # access cell
print(df.loc[["row2","row3"],["col1","col2"]])

#using colums
print(df[["col1", "col2"]])
```

```python
#using .iloc  we can access elements by indexing using .iloc

df.iloc[0]      #first row
print(df.iloc[0,2])   #first row and 3rd col
```

```python
#slicing in pandas
df.iloc[0:2, 2:4]
print(df.iloc[:,1:])
```
