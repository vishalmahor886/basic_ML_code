```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

Seaborn is a Python data visualization library built on top of Matplotlib.
It is mainly used to create beautiful, statistical graphics with less code.

Distribution Plot


*   displot
*   joinplot
*   pairplot





```python
df = sns.load_dataset('tips')
```

```python
df.head()
```

```python
df.corr(numeric_only=True)
```

```python
sns.heatmap(df.corr(numeric_only=True))
```

JOINT PLOTS

a joint plot is used to visual the relationship between two numerical variables

```python
sns.jointplot(data=df, x='tip', y='total_bill', kind='hex')
```

```python
sns.jointplot(data=df, x='tip', y='total_bill', kind='reg')
```

```python
sns.pairplot(df)
```

```python
sns.pairplot(df, hue='smoker')
```

```python
sns.distplot(df['total_bill'], kde=False, bins=10)
```

Categorical plots

used to visualize category variables

countplot
violinplot
boxplot
barplot

```python
#count plot
sns.countplot(data=df, x='day')
```

```python
#bar plots
sns.barplot(data=df, x='total_bill', y='sex')
```

```python
sns.boxplot(data=df,x='day',y='total_bill', palette='rainbow')
```

```python
sns.boxplot(data=df,x='day',y='total_bill', palette='rainbow', orient='h')
```

```python
sns.boxplot(data=df,x='total_bill',y='day', hue='smoker')
```

```python
sns.violinplot(data=df, x='total_bill', y='day')
```
